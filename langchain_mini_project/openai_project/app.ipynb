```python
from langchain_chroma import Chroma
from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEmbeddings

docs = [
    Document(
        page_content="Amitabh Bachchan is an Indian actor who primarily works in Hindi cinema. He made his film debut with Saat Hindustani in 1969 and has appeared in more than 200 films.",
        metadata={"source": "amitabh_bachchan.txt", "category": "Bollywood Actor"}
    ),

    Document(
        page_content="Shah Rukh Khan is an Indian actor and film producer known for his work in Hindi cinema. He made his film debut with Deewana in 1992 and has appeared in more than 100 films.",
        metadata={"source": "shah_rukh_khan.txt", "category": "Bollywood Actor"}
    ),

    Document(
        page_content="Aamir Khan is an Indian actor and filmmaker who works primarily in Hindi films. He is known for films such as Lagaan, Dil Chahta Hai, 3 Idiots, Dangal and PK.",
        metadata={"source": "aamir_khan.txt", "category": "Bollywood Actor"}
    ),

    Document(
        page_content="Salman Khan is an Indian actor and film producer who works primarily in Hindi cinema. Some of his well-known films include Hum Aapke Hain Koun, Bajrangi Bhaijaan, Sultan and Dabangg.",
        metadata={"source": "salman_khan.txt", "category": "Bollywood Actor"}
    ),

    Document(
        page_content="Ranbir Kapoor is an Indian actor who works in Hindi-language films. He made his acting debut with Saawariya in 2007 and is known for films such as Rockstar, Barfi!, Yeh Jawaani Hai Deewani and Sanju.",
        metadata={"source": "ranbir_kapoor.txt", "category": "Bollywood Actor"}
    ),

    Document(
        page_content="Rajkummar Rao is an Indian actor known for working in both independent and commercial Hindi films. He made his film debut with Love Sex Aur Dhokha in 2010 and is known for films such as Stree, Newton and Shahid.",
        metadata={"source": "rajkummar_rao.txt", "category": "Bollywood Actor"}
    )
]

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

vector_store = Chroma(
    embedding_function=embeddings,
    persist_directory="chroma_database",
    collection_name="sample_data"
)


## Basic step with chroma db (insertion , reading, updating and deleting)

## insert
# vector_store.add_documents(docs)
vector_store.add_documents(
    documents=docs,
    ids=[
        "amitabh_bachchan",
        "shah_rukh_khan",
        "aamir_khan",
        "salman_khan",
        "ranbir_kapoor",
        "rajkummar_rao"
    ]
)

## Reading


vector_store.get(include = ['documents', 'embeddings', 'metadatas'])

vector_store.similarity_search(
    query="Name of two action hero",
    k=2
)


# vector_store.update_document()





```

```python

updated_raj_doc = Document(
    page_content="Rajkummar Rao is an Indian actor known for working in both independent and commercial Hindi films. He made his film debut with Love Sex Aur Dhokha in 2010 and gained recognition through films such as Shahid, Newton, Stree, and Bareilly Ki Barfi. His recent films include Srikanth, Mr. & Mrs. Mahi, Stree 2, Vicky Vidya Ka Woh Wala Video, Maalik, and Toaster. His upcoming projects include Prahaar: The Untold Story of Ujjwal Nikam, Dada, Raftaar, and Vishkanya.",
    metadata={
        "source": "rajkummar_rao.txt",
        "category": "Bollywood Actor"
    }
)

data = vector_store.get()

print(data["ids"])

vector_store.update_documents(
    ids=["rajkummar_rao"],
    documents=[updated_raj_doc]
)

result = vector_store.get()

# print(data["documents"])
# print(data["metadatas"])


## Delete Operation
vector_store.delete(
    ids=["rajkummar_rao"]
)

vector_store.get(include = ['documents', 'embeddings', 'metadatas'])
```
