```python
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.prompts import ChatPromptTemplate
from langchain_groq import ChatGroq
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
load_dotenv()
```

```python
message_history = ChatMessageHistory()
message_history.add_user_message("Hello")
message_history.add_ai_message("Hii There")

# print(message_history)

store = {}

def get_history(session_id: str):
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]  

llm = ChatGroq(model = 'openai/gpt-oss-120b')
parser = StrOutputParser()
prompt = ChatPromptTemplate.from_messages([('system','You are an helpful Assitant'),('human', '{input}')])

chain =  prompt| llm | parser


chat_with_history = RunnableWithMessageHistory(
    chain, 
    get_history, 
    input_messages_key="input",
    output_messages_key="output"
    )

response = chat_with_history.invoke({"input":"Hello , Hii I am Vichitra , How are you"}, config={"configurable":{'session_id':'vichi@123'}})
print(response)

response_1 = chat_with_history.invoke({"input":"What is my name"}, config={"configurable":{'session_id':'vichi@123'}})
print(response_1)
```
