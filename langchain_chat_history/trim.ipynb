```python
from langchain_core.messages import SystemMessage, AIMessage, trim_messages, HumanMessage
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_groq import ChatGroq
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough
from operator import itemgetter
from dotenv import load_dotenv
load_dotenv()
```

```python
llm = ChatGroq(model = 'openai/gpt-oss-120b')
parser = StrOutputParser()
prompt = ChatPromptTemplate.from_messages([('system','You are an helpful Assitant'),('human', '{input}')])

trimmmer = trim_messages(max_tokens =30, token_counter = llm, include_system=True, allow_partial=False, start_on='human')

## some message for testing
messages = [
    HumanMessage(content = "Hello"),
    AIMessage(content = "Hii There"),
    HumanMessage(content="Tell me a story about AI"),
    AIMessage(content="Once Upon a time ..... very long response..."),
    HumanMessage(content="Can u summarize it"),
    AIMessage(content="Sure , I can do it so as ")
]

trimmmer.invoke(messages)


```
