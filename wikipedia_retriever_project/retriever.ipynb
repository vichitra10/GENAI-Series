```python
from langchain_community.vectorstores import Chroma
from langchain_huggingface import HuggingFaceEmbeddings
from sentence_transformers import SentenceTransformer
from langchain_core.documents import Document

model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
embeddings = HuggingFaceEmbeddings()

docs = [
    Document(
        page_content="Deep Learning is an important part of Generative AI and is used to train neural networks."
    ),
    Document(
        page_content="Generative AI uses deep learning models such as Transformers, GANs, and diffusion models to generate new content."
    ),
    Document(
        page_content="Large Language Models are a type of Generative AI that can understand and generate human-like text."
    )
]

vector_store = Chroma.from_documents(documents=docs, embedding= embeddings, collection_name='my_collection')
retriever = vector_store.as_retriever(search_kwargs={"k":2})

query = "How deep learning helps in acheiving in GENAI"
answer = retriever.invoke(query)
answer

for i,doc in enumerate(answer):
    print(f"Result:{i+1}")
    print(doc.page_content)
```

```python
from langchain_classic.retrievers import MultiQueryRetriever
from langchain_community.vectorstores import FAISS
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint, HuggingFaceEmbeddings
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from dotenv import load_dotenv
import os
from huggingface_hub import InferenceClient

load_dotenv()
HF_Token = os.getenv("HF_TOKEN")

llm = HuggingFaceEndpoint(
    repo_id="openai/gpt-oss-120b",
    huggingfacehub_api_token=HF_Token
)

chat_model = ChatHuggingFace(llm=llm)
embeddings = HuggingFaceEmbeddings()

docs = [
    "Deep Learning helps Generative AI by learning complex patterns and relationships from large amounts of data. These patterns allow AI models to understand existing information and generate new content.",

    "Deep Learning models such as Transformers, GANs, and diffusion models are important technologies behind Generative AI. They can be trained on large datasets and used to generate text, images, audio, videos, and other types of content.",

    "Generative AI uses Deep Learning to understand the structure and meaning of data. For example, language models learn from large collections of text and can use this knowledge to generate meaningful and human-like responses.",

    "Transformers are one of the most important Deep Learning architectures used in Generative AI. They use attention mechanisms to understand relationships between different words or tokens and help models generate relevant and coherent text.",

    "Deep Learning allows Generative AI models to improve their ability to generate content by learning from training data. During training, the model learns patterns, features, and relationships that can later be used to create new content.",

    "Large Language Models are built using Deep Learning techniques and are trained on huge amounts of text data. These models can understand questions, summarize information, translate languages, write content, and generate human-like conversations.",

    "GANs, or Generative Adversarial Networks, use two neural networks called the Generator and Discriminator. The Generator creates new data while the Discriminator evaluates whether the data is real or generated, helping the system produce more realistic results.",

    "Diffusion models are another type of Deep Learning model used in Generative AI, especially for image generation. They learn how to create new images by gradually removing noise and generating meaningful visual content.",

    "Overall, Deep Learning provides the foundation for many Generative AI systems. By learning complex patterns from large datasets, Deep Learning models enable Generative AI applications to create new and useful content."
]


query = "How Deep Learning Help in Achieving Generative AI?"
vector_store = FAISS.from_texts(docs, embeddings)
retriever = MultiQueryRetriever.from_llm(
    retriever=vector_store.as_retriever(
        search_kwargs={"k": 3}
    ),
    llm=chat_model
)

result = retriever.invoke(query)
for i, doc in enumerate(result):
    print(f"Result {i + 1}:")
    print(doc.page_content)
    print()

```

```python
from langchain_classic.retrievers import ContextualCompressionRetriever
from langchain_classic.retrievers.document_compressors import LLMChainExtractor


docs = [
    "Python lists are ordered and mutable collections. You can add, remove, and modify elements in a list. Lists are created using square brackets, such as [10, 20, 30].",

    "Python dictionaries store data in key-value pairs. They are useful when you need to access a value using a specific key. Dictionaries are created using curly brackets.",

    "Python functions are reusable blocks of code designed to perform a particular task. Functions are defined using the def keyword and can accept parameters and return values.",

    "Python lists support many useful methods such as append(), extend(), insert(), remove(), pop(), sort(), and reverse(). These methods allow developers to modify list contents.",

    "Python classes are used to implement object-oriented programming. A class can contain attributes and methods and can be used to create multiple objects.",

    "Python tuples are ordered collections that cannot be changed after creation. Tuples are generally written using parentheses, such as (10, 20, 30).",

    "Python list comprehensions provide a concise way to create lists. For example, [x * 2 for x in numbers] creates a new list containing twice the value of each element."
]

query = "What are the important features and methods of Python lists?"

vector_store = FAISS.from_texts(docs, embeddings)

base_retriever = vector_store.as_retriever(
    search_kwargs={"k": 3}
)

compressor = LLMChainExtractor.from_llm(chat_model)

compression_retriever = ContextualCompressionRetriever(
    base_retriever=base_retriever,
    base_compressor=compressor
)

result = compression_retriever.invoke(query)

for i, doc in enumerate(result):
    print(f"Result {i + 1}:")
    print(doc.page_content)
    print()
```
