## Q.1: What are SystmeMessage, HumanMessage and AIMessage ?
## Ans: SystemMessage: It tells the LLM how it should behave and respond ----- (Instructions)
## HumanMessage: It represent the message sent by the user ---- (User's Question)
## AIMessage: It represents the response generate by the LLM ----(AI's response)

## Q.2: Why message-based prompting is better than single prompts?
## Ans: In LangChain, message based prompting is often better than a single prompt because it seprates the instructions, user questions and AI response into clear roles. This is especially useful when building chatbots and managing conversation history


## Task 2: MessagePlaceHolder Usage

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import SystemMessage, HumanMessage , AIMessage
from langchain_groq import ChatGroq
from dotenv import load_dotenv

load_dotenv()

## Initialize the Model
model = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0
)

## creating Prompt Template
prompts = ChatPromptTemplate([
    ('system','You are a helpful Chatbot. Use the conversation history to answer naturally'),
    MessagesPlaceholder(variable_name='history'),
    ('human', '{input}')
])

history = [
    HumanMessage(content= "My name is Vichita"),
    AIMessage(content="Nice to meet you , Vichitra!"),
    HumanMessage(content="I am learning Langchain"),
    AIMessage(content="That's great , Langchain most demanding now a days")
]


chain =  prompts | model

response = chain.invoke({
    "history": history,
    "input": "What programming language am I learning?"
})

print(response.content)

response1 = chain.invoke({
    "history": history,
    "input": "Can you tell me what is my name?"
})

print(response1.content)
```

## Part 2 : Conversation History Management
## Task 3: Basic Message History

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import SystemMessage, HumanMessage , AIMessage
from langchain_groq import ChatGroq
from dotenv import load_dotenv

load_dotenv()

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0
)

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful chatbot. Use the conversation history to answer naturally."),
    MessagesPlaceholder(variable_name="history")
])

chain = prompt | llm

history = []
while True:
    user_input = input("You: ")
    if user_input.lower() == 'exit':
        break
    
    history.append(HumanMessage(content = user_input))

response = chain.invoke({"history": history})
print("AI:", response.content)
history.append(AIMessage(content=response.content))


```
