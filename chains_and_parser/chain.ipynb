```python
from langchain_groq import ChatGroq
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
import google.generativeai as genai
import torch
import os

load_dotenv()
```

```python
## Simple Sequential Chain

prompt = PromptTemplate(template = 'Create a 5 points summary on {Topic}', input_variable=['Topic'])

google_model = ChatGoogleGenerativeAI(model = 'gemini-3.5-flash')
groq_chat_model = ChatGroq(model="openai/gpt-oss-120b",temperature=0)

parser = StrOutputParser()

chain = prompt | groq_chat_model | parser
result = chain.invoke({'Topic': 'Cricket Asia Cup'})
result
```

```python
prompt_1 = PromptTemplate(template= 'Create a Detailed Report on {Topic}', input_variable=['Topic'])
model_1= google_model

prompt_2 = PromptTemplate(template = 'Create a 5 point summary on the {Report}', input_variable = ['Report'])
model_2 = ChatGroq(model="openai/gpt-oss-120b",temperature=0)

parser = StrOutputParser()

chain = prompt_1 | model_1 | parser | prompt_2 | model_2 | parser
output = chain.invoke({'Topic': 'Cricket World Cup'})

print(output)
```
