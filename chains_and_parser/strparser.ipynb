```python
from langchain_groq import ChatGroq
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
import google.generativeai as genai
import torch
import os
```

```python
load_dotenv()
```

```python
google_key = os.getenv('GOOGLE_API_kEY')
genai.configure(api_key = google_key)
```

```python
g_model = genai.GenerativeModel("gemini-3.8-flash")
response = g_model.generate_content('Hello')
response
```

```python
google_model = ChatGoogleGenerativeAI(model = 'gemini-3.5-flash')
groq_chat_model = ChatGroq(model="openai/gpt-oss-120b",temperature=0)


prompt1 = PromptTemplate(template="""Generate a detailed report on the {topic}""", input_variables = ['topic'])
prompt2 = PromptTemplate(template="""Create a 5 point on summary on this {report}""", input_variables = ['report'])


prompt1.invoke({'topic': 'Google vs Groq api'})

parser = StrOutputParser()


chain = prompt1 | google_model | parser | prompt2 | groq_chat_model | parser
response = chain.invoke({'topic': 'Google vs Groq api'})
response
```
