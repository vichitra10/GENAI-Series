```python
from langchain_groq import ChatGroq
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
import google.generativeai as genai
from langchain_core.runnables import RunnableParallel, RunnableSequence, RunnableLambda, RunnablePassthrough
import torch
import os

load_dotenv()
```

```python
def wordN(words):
    return len(words.split())
```

```python
prompt = PromptTemplate(template='Generate a joke on this {Topic}', input_variables=['Topic'])

model = ChatGroq(model="openai/gpt-oss-120b",temperature=1)
parser = StrOutputParser()

chain = RunnableSequence(prompt,model,parser)
parallel_chain = RunnableParallel({'joke': RunnablePassthrough(), 'Words Count': RunnableLambda(wordN) })
final_chain = chain|parallel_chain
response = final_chain.invoke({'Topic': 'Robot and Human'})
response

```
