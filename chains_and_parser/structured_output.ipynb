```python
from pydantic import BaseModel, Field
from typing import Optional, List, Literal
from langchain_groq import ChatGroq
import torch

from dotenv import load_dotenv
load_dotenv()
```

```python
class Review(BaseModel):
    Product: str = "No Info"
    Pros : List[str]
    Cons : List[str]
    Reviewer: Optional[str]=None
    ratings: float = Field(
        gt=0,
        lt=5,
        default=3,
        description="Return Rating of the Reviewer for the Current Product"
    )  
```

```python
class Data(BaseModel):

    Product: str = Field(
        description="Provide the name of the product"
    )

    Pros: list[str] = Field(
        description="Mention Positive Points mentioned about the product"
    )

    Cons: list[str] = Field(
        description="Mention the Negative Points mentioned about the product"
    )

    Sentiments: Literal["Positive", "Negative"] = Field(
        description="Mention the Sentiment of the user review"
    )

    Ratings: float = Field(
        ge=1,
        le=5,
        default=3,
        description="Rating provided by the user as per the review"
    )

    Reviewer: Optional[str] = Field(
        description="Name of the Reviewer",
        default=None
    )


# OUTSIDE the Data class
review = """
I am very disappointed with this product. The sound quality is poor,
the battery drains quickly, and the build quality feels cheap.
Considering its price, I expected much better performance.
I would give this product a rating of 2 out of 5 and would not recommend it.
"""

model_id = "openai/gpt-oss-120b"

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0
)

structured_llm = llm.with_structured_output(Data)

response = structured_llm.invoke(review)

print(response)
```
